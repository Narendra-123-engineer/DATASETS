```python

import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
from sklearn.metrics import confusion_matrix,accuracy_score

#load inbuilt Iris dataset
iris=load_iris()
x=iris.data
y=iris.target #actual labels

#apply k means (3 clusters for 3 flower types)
kmeans=KMeans(n_clusters=3,random_state=42)
y_kmeans=kmeans.fit_predict(x)

#create dataframe for results
df = pd.DataFrame(x, columns=iris.feature_names)
df['Actual'] = y
df['Cluster'] = y_kmeans

print("cluster centers:\n", kmeans.cluster_centers_)
print("\nsample clustered data;\n",df.head())

#evaluate clustering(note;cluster lables may not match true labels directly)
cm = confusion_matrix(y,y_kmeans)
print("\nconfusion matrix;\n", cm)

#accuracy(after lablealignment,not raw)
#since cluster numbers don't correspond to actual labels ,accuracy may look low without relabeling.
```
