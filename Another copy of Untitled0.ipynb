```python
x=int(input("enter first number"))
y=int(input("enter second number"))
op=input("enter the operator("+,-,*,/")")
if op=="+":
  print(x+y)
elif op=="-":
  print(x-y)
elif op=="*":
  print(x*y)
elif op=="/":
  print(x/y)
```
